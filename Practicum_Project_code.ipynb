```python
import requests
import pandas as pd

api_key = "64da1632f6a904859a1283a48272a1c0"


def fetch_fred_series(series_id, api_key):
    url = f"https://api.stlouisfed.org/fred/series/observations"
    params = {
        "series_id": series_id,
        "api_key": api_key,
        "file_type": "json"
    }
    r = requests.get(url, params=params)
    obs = r.json()["observations"]
    df = pd.DataFrame(obs)
    df["date"] = pd.to_datetime(df["date"])
    df = df[["date", "value"]]
    df["value"] = pd.to_numeric(df["value"], errors="coerce")
    return df
```

```python
quarterly_series_dict = {
    "NominalGDP": "GDP",
    "RealGDP": "GDPC1",
    "GrossDomesticProduct": "GDP",
    "CorporateProfits": "CP",
    "DisposableIncome": "DPI",
    "GrossPrivateDomesticInvestment": "GPDI",
    "ProducerPriceIndexPolyesterFibers": "WPU031502411",
    "PersonalConsumptionExp_Quarterly": "PCECC96",
    "LaborProductivity": "OPHNFB"
}

def fetch_fred_series(series_id, api_key):
    url = f"https://api.stlouisfed.org/fred/series/observations"
    params = {
        "series_id": series_id,
        "api_key": api_key,
        "file_type": "json"
    }
    r = requests.get(url, params=params)
    obs = r.json()["observations"]
    df = pd.DataFrame(obs)
    df["date"] = pd.to_datetime(df["date"])
    df = df[["date", "value"]]
    df["value"] = pd.to_numeric(df["value"], errors="coerce")
    df["quarter"] = df["date"].dt.to_period("Q")
    df = df.drop(columns=["date"])
    return df

main_quarterly_df = None
for col_name, series_id in quarterly_series_dict.items():
    print(f"Fetching quarterly: {col_name}")
    temp_df = fetch_fred_series(series_id, api_key).rename(columns={"value": col_name})
    if main_quarterly_df is None:
        main_quarterly_df = temp_df
    else:
        main_quarterly_df = pd.merge(main_quarterly_df, temp_df, on="quarter", how="outer")

main_quarterly_df = main_quarterly_df[(main_quarterly_df["quarter"] >= pd.Period("2020Q1", "Q")) & (main_quarterly_df["quarter"] <= pd.Period("2024Q4", "Q"))]
main_quarterly_df = main_quarterly_df.sort_values("quarter").reset_index(drop=True)
main_quarterly_df = main_quarterly_df.fillna(method="ffill")

print(main_quarterly_df.shape)
print(main_quarterly_df.isna().sum())
```

# Add Inflation

InflationRate=(CurrentCPI−PreviousCPI)/PreviousCPI×100%

```python
df = final_merged_df.copy()
monthly_cpi = (
    df[['month', 'CPI']]
    .drop_duplicates(subset=['month'])
    .sort_values('month')
    .reset_index(drop=True)
)

monthly_cpi['PreviousCPI'] = monthly_cpi['CPI'].shift(1)
monthly_cpi['InflationRate'] = ((monthly_cpi['CPI'] - monthly_cpi['PreviousCPI']) / monthly_cpi['PreviousCPI']) * 100

df = pd.merge(df, monthly_cpi[['month', 'InflationRate']], on='month', how='left')

df['InflationRate'] = df['InflationRate'].fillna(0)

print(df[['INVOICE DATE', 'month', 'CPI', 'InflationRate']])
```

# USA season


Winter: December, January, February (int = 1)

Spring: March, April, May (int = 2)

Summer: June, July, August (int = 3)

Fall (Autumn): September, October, November (int = 4)

```python
def month_to_season(month):
    if month in [12, 1, 2]:
        return 1   # Winter
    elif month in [3, 4, 5]:
        return 2   # Spring
    elif month in [6, 7, 8]:
        return 3   # Summer
    elif month in [9, 10, 11]:
        return 4   # Fall

df["Season"] = df["INVOICE DATE"].dt.month.apply(month_to_season)
df
```

## Imports_Vietnam – IMP5520
- **Ticker**: IMP5520
- **Name**: U.S. Imports of Goods by Customs Basis from Vietnam.
- **Description**: Monthly value (millions of dollars, not seasonally adjusted) of all goods the United States imports from Vietnam, measured on a customs‑basis. It approximates how much product the U.S. is sourcing from Vietnam and can proxy supply exposure to Vietnamese manufacturing.

## Imports_China – IMPCH
- **Ticker**: IMPCH
- **Name**: U.S. Imports of Goods by Customs Basis from China.
- **Description**: Monthly customs‑basis imports of goods from China into the U.S., in millions of dollars, not seasonally adjusted. This series tracks the scale of U.S. sourcing from China and can be related to Nike’s reliance on Chinese production or broader supply‑chain conditions.

## FX_IDR_USD_EOP – CCUSSP02IDM650N
- **Ticker**: CCUSSP02IDM650N
- **Name**: Currency Conversions: US Dollar Exchange Rate: Spot, End of Period: National Currency: USD for Indonesia.
- **Description**: Monthly end‑of‑period spot exchange rate between the U.S. dollar and the Indonesian rupiah (national currency per 1 USD). It reflects how many rupiah equal one dollar at the end of each month and captures FX cost conditions for products manufactured in Indonesia.

## FX_CNY_USD – DEXCHUS
- **Ticker**: DEXCHUS
- **Name**: Chinese Yuan Renminbi to U.S. Dollar Spot Exchange Rate.
- **Description**: Daily spot exchange rate showing how many Chinese yuan renminbi equal one U.S. dollar, not seasonally adjusted. It measures currency movements between China and the U.S., relevant for Nike’s China‑linked production costs and U.S. margins.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

!pip install pmdarima
import pmdarima as pm
```

```python
weekly = df.set_index('INVOICE DATE').groupby('id_Region').resample('W')['Units Sold'].sum().reset_index()
weekly = weekly.rename(columns={'INVOICE DATE': 'week', 'id_Region': 'region', 'Units Sold': 'units'})

regions = df['id_Region'].unique()
TEST_WEEKS = 52
ORIG_DRIVERS = ['Price per Unit', 'Costo Unitario', 'Operating Margin']

# evaluation metrics
def wmape(actual, prediction):
    return np.mean(np.abs(actual - prediction)) / np.mean(actual) * 100

def rmse(actual, prediction):
    return np.sqrt(np.mean((actual - prediction)**2))

def mae(actual, prediction):
    return np.mean(np.abs(actual - prediction))

def r2(actual, prediction):
    actual, prediction = np.asarray(actual, float), np.asarray(prediction, float)
    ss_res = np.sum((actual - prediction) ** 2)
    ss_tot = np.sum((actual - np.mean(actual)) ** 2)
    return 1 - ss_res / (ss_tot + 1e-9)


def fourier_terms(t, period, K):
    cols = []
    for k in range(1, K + 1):
        cols.append(np.sin(2 * np.pi * k * t / period))
        cols.append(np.cos(2 * np.pi * k * t / period))
    return np.column_stack(cols)

snaive_a, snaive_p = [], []
sarimax_a, sarimax_p = [], []
per_region_rows = []

for reg in regions:
    sub = weekly[weekly["region"] == reg].sort_values("week").reset_index(drop=True)
    y   = sub["units"].values.astype(float)
    n   = len(y)

    if n < TEST_WEEKS + 52:
        print(f"  [{reg}] Not enough data for region {reg}. Skipping.")
        continue

    tr_end = n - TEST_WEEKS
    y_tr, y_te = y[:tr_end], y[tr_end:]

    # seasonal naive: value from 52 weeks earlier (fallback: last train value)
    snaive = []
    for h in range(TEST_WEEKS):
        idx = tr_end + h - 52
        snaive.append(y[idx] if (idx >= 0 and idx < n) else y_tr[-1])
    snaive = np.clip(np.array(snaive), 0, None)

    #  SARIMAX: ORIGINAL drivers only (price/cost/margin)
    t   = np.arange(n)
    F   = fourier_terms(t, 52.18, 2)
    Xo  = sub[ORIG_DRIVERS].values.astype(float) if all(driver in sub.columns for driver in ORIG_DRIVERS) else np.empty((n, 0))
    X   = np.hstack([Xo, F])
    X_tr, X_te = X[:tr_end], X[tr_end:]

    # Handle case where X_tr or y_tr is empty for auto_arima
    if len(y_tr) == 0 or len(X_tr) == 0:
        print(f"  [{reg}] Not enough training data for SARIMAX. Using seasonal-naive.")
        fc = snaive
    else:
        try:
            m = pm.auto_arima(y_tr, X=X_tr, seasonal=False,
                              start_p=0, max_p=3, start_q=0, max_q=3, d=None,
                              information_criterion="aic", stepwise=True,
                              suppress_warnings=True, error_action="ignore")
            fc = np.clip(m.predict(n_periods=TEST_WEEKS, X=X_te), 0, None)
        except Exception as e:
            print(f"  [{reg}] SARIMAX failed ({e}); using seasonal-naive")
            fc = snaive

    snaive_a.append(y_te);  snaive_p.append(snaive)
    sarimax_a.append(y_te); sarimax_p.append(fc)
    per_region_rows.append({"region": reg,
                            "SARIMAX_WMAPE": round(wmape(y_te, fc), 2),
                            "SNaive_WMAPE":  round(wmape(y_te, snaive), 2)})

snaive_a  = np.concatenate(snaive_a);  snaive_p  = np.concatenate(snaive_p)
sarimax_a = np.concatenate(sarimax_a); sarimax_p = np.concatenate(sarimax_p)

baseline_results = {
    "Model": "SARIMAX (orig only)",
    "RMSE":  round(rmse(sarimax_a, sarimax_p), 2),
    "MAE":   round(mae(sarimax_a,  sarimax_p), 2),
    "WMAPE": round(wmape(sarimax_a, sarimax_p), 2),
    "R2":    round(r2(sarimax_a, sarimax_p), 3),
}
snaive_results = {
    "Model": "Seasonal-naive",
    "RMSE":  round(rmse(snaive_a, snaive_p), 2),
    "MAE":   round(mae(snaive_a,  snaive_p), 2),
    "WMAPE": round(wmape(snaive_a, snaive_p), 2),
    "R2":    round(r2(snaive_a, snaive_p), 3),
}
print("\n── BASELINE (weekly, region-wise, original columns only) ──")
print(pd.DataFrame(per_region_rows).to_string(index=False))
print("\nSARIMAX       :", baseline_results)
print("Seasonal-naive:", snaive_results)
```

```python
import warnings; warnings.filterwarnings("ignore")
import contextlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import lightning.pytorch as pl
from lightning.pytorch.callbacks import EarlyStopping, LearningRateMonitor, ModelCheckpoint
from lightning.pytorch.loggers import CSVLogger
from pytorch_forecasting import TimeSeriesDataSet, TemporalFusionTransformer, Baseline
from pytorch_forecasting.data import GroupNormalizer, NaNLabelEncoder
from pytorch_forecasting.metrics import MAE, QuantileLoss
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

pl.seed_everything(42, workers=True)
torch.set_float32_matmul_precision("medium")
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"PyTorch {torch.__version__} | accelerator: {DEVICE}")
if DEVICE == "cpu":
    print("NOTE: on CPU. In Colab enable GPU (Runtime → Change runtime type → GPU) "
          "for much faster Optuna.")

# metric helpers
def rmse(a, p):  return float(np.sqrt(mean_squared_error(a, p)))
def mae(a, p):   return float(mean_absolute_error(a, p))
def r2(a, p):    return float(r2_score(a, p))
def wmape(a, p):
    a, p = np.asarray(a, float), np.asarray(p, float)
    return 100 * np.sum(np.abs(a - p)) / (np.sum(np.abs(a)) + 1e-9)

try:
    import pytorch_forecasting.data.encoders as _pfe
    torch.serialization.add_safe_globals(
        [getattr(_pfe, n) for n in
         ["GroupNormalizer", "TorchNormalizer", "EncoderNormalizer",
          "NaNLabelEncoder", "MultiNormalizer"] if hasattr(_pfe, n)])
except Exception as _e:
    print("add_safe_globals skipped:", _e)

@contextlib.contextmanager
def trusted_torch_load():
    """Force full (trusted) unpickling for our own checkpoints."""
    _orig = torch.load
    def _tl(*a, **k): k["weights_only"] = False; return _orig(*a, **k)
    torch.load = _tl
    try:
        yield
    finally:
        torch.load = _orig

TEST_WEEKS = 8       # forecast horizon
VAL_WEEKS  = 8       # validation window
REGION_MAP = {1: "Northeast", 2: "South", 3: "West", 4: "Midwest", 5: "Southeast"}
```

```python
# Build WEEKLY region-wise table
FRED_CANDIDATES = [
    "SP500", "USDollarIndex", "FX_CNY_USD", "CPI", "InflationRate",
    "UnemploymentRate", "RetailSalesTotal", "RetailSalesClothing",
    "RetailSalesSportingGoodsStores", "PCE", "FederalFundsRate", "SavingsRate",
    "SentimentIndex", "AvgWeeklyEarnings", "AvgHourlyEarningsRetail",
    "Population", "PPI", "PPI_Footwear", "BusinessInventories",
    "EmploymentNonfarm", "LaborForceParticipation", "ConsumerCreditOutstanding",
    "HousingStarts", "Imports_Vietnam", "Imports_China", "FX_IDR_USD_EOP",
    "NominalGDP", "RealGDP", "GrossDomesticProduct", "CorporateProfits",
    "DisposableIncome", "GrossPrivateDomesticInvestment",
    "ProducerPriceIndexPolyesterFibers", "PersonalConsumptionExp_Quarterly",
    "LaborProductivity","RegionalCPI",
]
FRED_PRESENT = [c for c in FRED_CANDIDATES if c in df.columns]
ORIG_DRIVERS = ["price", "cost", "margin"]

d = df.copy()
d["INVOICE DATE"] = pd.to_datetime(d["INVOICE DATE"])
d["week"] = d["INVOICE DATE"].dt.to_period("W-SUN").dt.start_time

weekly = (
    d.groupby(["week", "id_Region"])
    .agg(
        units  = ("Units Sold",       "sum"),
        price  = ("Price per Unit",   "mean"),
        cost   = ("Costo Unitario",   "mean"),
        margin = ("Operating Margin", "mean"),
        season = ("Season",           "first"),
        **{c: (c, "mean") for c in FRED_PRESENT},
    )
    .reset_index()
    .rename(columns={"id_Region": "region"})
)
weekly["region"] = weekly["region"].map(REGION_MAP).fillna(weekly["region"].astype(str))

regions   = weekly["region"].unique().tolist()
all_weeks = pd.date_range(weekly["week"].min(), weekly["week"].max(), freq="W-MON")
grid = (pd.MultiIndex.from_product([regions, all_weeks], names=["region", "week"])
        .to_frame(index=False))
weekly = grid.merge(weekly, on=["region", "week"], how="left") \
             .sort_values(["region", "week"]).reset_index(drop=True)

weekly["units"] = weekly["units"].fillna(0)
for c in ORIG_DRIVERS + FRED_PRESENT + ["season"]:
    weekly[c] = weekly.groupby("region")[c].ffill().bfill()
print(f"Weekly table: {weekly.shape} | regions: {len(regions)} | weeks/region: {len(all_weeks)}")
```

```python
# Feature selection (rank ALL FRED + original by |corr|)
_tot = (weekly.groupby("week")
        .agg(units=("units", "sum"),
             **{c: (c, "mean") for c in FRED_PRESENT + ORIG_DRIVERS})
        .reset_index())
for c in FRED_PRESENT + ORIG_DRIVERS:
    _tot[c] = pd.to_numeric(_tot[c], errors="coerce")
_tot = _tot.ffill().bfill()

_cands = FRED_PRESENT + ORIG_DRIVERS
_rank  = (_tot[["units"] + _cands].corr()["units"].drop("units")
          .abs().sort_values(ascending=False))
_cmat  = _tot[_cands].corr().abs()

TOP_K = 10
SELECTED = []
for cand in _rank.index:
    if len(SELECTED) >= TOP_K:
        break
    if all(_cmat.loc[cand, s] < 0.90 for s in SELECTED):
        SELECTED.append(cand)

SEL_FRED = [c for c in SELECTED if c in FRED_PRESENT]
SEL_ORIG = [c for c in SELECTED if c in ORIG_DRIVERS]
print("Selected features (by |corr| w/ weekly demand):")
print(_rank[SELECTED].round(3).to_string())
print(f"\n  known (FRED)   : {SEL_FRED}")
print(f"  unknown (orig) : {SEL_ORIG}")
```

```python
# Weekly feature engineering (EWMA + target history + calendar)
panel = weekly.copy()
GKEY  = "region"
gp    = panel.groupby(GKEY, group_keys=False)

FRED_EXP = []
for col in SEL_FRED:
    for span in [4, 12]:
        nm = f"{col}_ewm{span}"
        panel[nm] = gp[col].apply(lambda s: s.ewm(span=span, adjust=False).mean())
        FRED_EXP.append(nm)

TARGET_HIST = []
for lag in [1, 4, 12, 52]:
    nm = f"units_lag{lag}"; panel[nm] = gp["units"].shift(lag); TARGET_HIST.append(nm)
for span in [4, 12]:
    nm = f"units_ewm{span}"
    panel[nm] = gp["units"].apply(lambda s: s.shift(1).ewm(span=span, adjust=False).mean())
    TARGET_HIST.append(nm)
for w in [4, 12]:
    mnm, snm = f"units_roll{w}_mean", f"units_roll{w}_std"
    panel[mnm] = gp["units"].apply(lambda s: s.shift(1).rolling(w, min_periods=1).mean())
    panel[snm] = gp["units"].apply(lambda s: s.shift(1).rolling(w, min_periods=1).std())
    TARGET_HIST += [mnm, snm]
panel[TARGET_HIST] = panel.groupby(GKEY)[TARGET_HIST].bfill().fillna(0)

panel["week_of_year"] = panel["week"].dt.isocalendar().week.astype(int)
panel["month"]        = panel["week"].dt.month
panel["quarter"]      = panel["week"].dt.quarter
panel["year"]         = panel["week"].dt.year
two_pi = 2 * np.pi
panel["woy_sin"]   = np.sin(two_pi * panel["week_of_year"] / 52.18)
panel["woy_cos"]   = np.cos(two_pi * panel["week_of_year"] / 52.18)
panel["month_sin"] = np.sin(two_pi * panel["month"] / 12)
panel["month_cos"] = np.cos(two_pi * panel["month"] / 12)
CYCLICAL = ["woy_sin", "woy_cos", "month_sin", "month_cos"]

panel = panel.sort_values([GKEY, "week"]).reset_index(drop=True)
panel["time_idx"] = panel.groupby(GKEY).cumcount()
panel["season"]   = panel["season"].astype(int).astype(str)
print(f"TFT panel: {panel.shape} | +{len(FRED_EXP)} EWMA +{len(TARGET_HIST)} target-history")
```

```python
weeks_per_region = int(panel.groupby(GKEY)["time_idx"].max().min()) + 1
#ENCODER_WEEKS = 104
ENCODER_WEEKS = 52
PREDICT_WEEKS = TEST_WEEKS
VAL_WEEKS     = TEST_WEEKS
print(f"weeks/region: {weeks_per_region} | encoder: {ENCODER_WEEKS} | horizon: {PREDICT_WEEKS}")

KNOWN_REALS   = (["time_idx", "week_of_year", "month", "quarter", "year"]
                 + CYCLICAL + SEL_FRED + FRED_EXP)
UNKNOWN_REALS = SEL_ORIG + TARGET_HIST
STATIC_CATS   = ("region")
TIME_CAT      = ["season"]

max_ti      = panel["time_idx"].max()
test_cutoff = max_ti - PREDICT_WEEKS
val_cutoff  = test_cutoff - VAL_WEEKS
train_df = panel[panel["time_idx"] <= val_cutoff].copy()
val_df   = panel[panel["time_idx"] <= test_cutoff].copy()

training_dataset = TimeSeriesDataSet(
    train_df,
    time_idx              = "time_idx",
    target                = "units",
    group_ids             = [GKEY],
    min_encoder_length    = ENCODER_WEEKS // 2,
    max_encoder_length    = ENCODER_WEEKS,
    min_prediction_length = 1,
    max_prediction_length = PREDICT_WEEKS,
    static_categoricals               = STATIC_CATS,
    time_varying_known_categoricals   = TIME_CAT,
    time_varying_known_reals          = KNOWN_REALS,
    time_varying_unknown_categoricals = [],
    time_varying_unknown_reals        = UNKNOWN_REALS + ["units"],
    target_normalizer = GroupNormalizer(groups=[GKEY], transformation="log1p"),
    categorical_encoders = {"region": NaNLabelEncoder(add_nan=True),
                            "season": NaNLabelEncoder(add_nan=True)},
    add_relative_time_idx   = True,
    add_target_scales       = True,
    add_encoder_length      = True,
    allow_missing_timesteps = True,
)
validation_dataset = TimeSeriesDataSet.from_dataset(
    training_dataset, val_df, predict=True, stop_randomization=True)

BATCH = 64
train_loader = training_dataset.to_dataloader(train=True,  batch_size=BATCH,     num_workers=0, shuffle=True)
val_loader   = validation_dataset.to_dataloader(train=False, batch_size=BATCH * 2, num_workers=0)

mae_metric = MAE().to(DEVICE)
base_pred = Baseline().predict(val_loader, return_y=True, trainer_kwargs={"accelerator": DEVICE})
print(f"Naive (last-value) MAE on val: {mae_metric(base_pred.output.to(DEVICE), base_pred.y[0].to(DEVICE)):.2f}")
```

# INVENTORY  —  region-level order-up-to (R,S) from the TFT

```python
from collections import defaultdict

REVIEW_WEEKS, LEAD_WEEKS = 1, 1
PROTECTION = REVIEW_WEEKS + LEAD_WEEKS
SERVICE_LEVEL = 0.95
Z = {0.90: 1.2816, 0.95: 1.6449, 0.98: 2.0537}[SERVICE_LEVEL]
HOLDING_RATE, ORDERING_COST = 0.25, 50.0
REGION_MAP = {1: "Northeast", 2: "South", 3: "West", 4: "Midwest", 5: "Southeast"}
_NAME2ID = {v: k for k, v in REGION_MAP.items()}
Q_WIDTH_TO_SIGMA = 1.0 / (2 * 1.2816)
```

```python
# TFT forecast: median + P10/P90 quantiles on the test window
pred = best_tft.predict(val_loader, mode="raw", return_x=True,
                        trainer_kwargs={"accelerator": "cpu"})
qs   = best_tft.loss.quantiles
i10, i50, i90 = qs.index(0.1), qs.index(0.5), qs.index(0.9)
QP   = pred.output.prediction.cpu().numpy()
p50  = np.clip(QP[..., i50], 0, None)
p10  = np.clip(QP[..., i10], 0, None)
p90  = np.clip(QP[..., i90], 0, None)
actuals = np.concatenate([y[0].cpu().numpy() for _, y in val_loader], axis=0)
H = p50.shape[1]
```

```python
# per-region cost inputs from data
_costs = df.groupby("id_Region").agg(unit_cost=("Costo Unitario", "mean"),
                                     price=("Price per Unit", "mean"))
# historical weekly demand std per region (for the "old way" comparison)
_d = df.copy(); _d["week"] = pd.to_datetime(_d["INVOICE DATE"]).dt.to_period("W-SUN").dt.start_time
_hist = (_d.groupby(["week", "id_Region"])["Units Sold"].sum().reset_index())
_hist_sig = _hist.groupby("id_Region")["Units Sold"].apply(lambda s: s.tail(52).std())
```

```python
# order-up-to simulator (R,S)
def simulate_RS(demand, S, R, L, holding_wk, penalty):
    on_hand, on_order = float(S), 0.0
    arr = defaultdict(float); hold = short = orders = tot = 0.0
    for t in range(len(demand)):
        on_hand += arr.get(t, 0.0); on_order -= arr.get(t, 0.0)
        d = demand[t]; tot += d
        sold = min(on_hand, d); sh = d - sold; on_hand -= sold; short += sh
        if t % R == 0:
            q = max(0.0, S - (on_hand + on_order))
            if q > 0: on_order += q; arr[t + L] += q; orders += 1
        hold += on_hand * holding_wk
    return dict(fill=1 - short / (tot + 1e-9), short=short, orders=orders,
                cost=hold * 1 + sh * 0 + short * penalty + orders * ORDERING_COST,
                holding=hold, stockout=short * penalty)
```

```python
# Build the FORECAST-DRIVEN policy per region and simulate
rows = []
for i, reg in enumerate(regions):
    rid = _NAME2ID.get(reg)
    unit_cost = float(_costs.loc[rid, "unit_cost"]); price = float(_costs.loc[rid, "price"])
    holding_wk = HOLDING_RATE * unit_cost / 52.0
    penalty    = max(price - unit_cost, 0.0)

    mu_fc  = float(p50[i].mean())
    sig_fc = float(np.mean((p90[i] - p10[i]) * Q_WIDTH_TO_SIGMA))
    sig_hist = float(_hist_sig.get(rid, sig_fc))
    def make_S(sig):
        ss = Z * sig * np.sqrt(PROTECTION)
        return round(ss), round(mu_fc * PROTECTION + ss)

    ss_fc, S_fc     = make_S(sig_fc)
    ss_hist, S_hist = make_S(sig_hist)

    res = simulate_RS(actuals[i], S_fc, REVIEW_WEEKS, LEAD_WEEKS, holding_wk, penalty)
    rows.append({"region": reg, "mu_fc": round(mu_fc), "sigma_fc": round(sig_fc),
                 "sigma_hist": round(sig_hist), "S(forecast)": S_fc, "S(hist)": S_hist,
                 "safety_fc": ss_fc, "safety_hist": ss_hist,
                 "fill_%": round(res["fill"] * 100, 1),
                 "cost_$": round(res["holding"] + res["stockout"] + res["orders"] * ORDERING_COST)})
inv = pd.DataFrame(rows)
```

```python
# Report
print("── FORECAST-DRIVEN order-up-to policy (safety stock from TFT quantiles) ──")
print(inv[["region", "mu_fc", "sigma_fc", "safety_fc", "S(forecast)", "fill_%", "cost_$"]].to_string(index=False))
print(f"\nPortfolio fill rate : {inv['fill_%'].mean():.1f}%   (target {SERVICE_LEVEL:.0%})")
print(f"Total cost          : ${inv['cost_$'].sum():,.0f}")

print("\n── Value of the forecast:  forecast-sigma vs historical-sigma safety stock ──")
print(inv[["region", "sigma_fc", "sigma_hist", "safety_fc", "safety_hist"]].to_string(index=False))
_red = 100 * (1 - inv["safety_fc"].sum() / max(inv["safety_hist"].sum(), 1))
print(f"\nTotal safety stock — forecast-driven: {inv['safety_fc'].sum():,.0f} units | "
      f"historical: {inv['safety_hist'].sum():,.0f} units")
print(f"=> The TFT's sharper uncertainty {'reduces' if _red>0 else 'changes'} required "
      f"safety stock by ~{abs(_red):.0f}% at the same 95% service level.")
```
